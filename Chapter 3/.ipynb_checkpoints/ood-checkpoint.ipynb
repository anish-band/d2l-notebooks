```python
import time
import numpy as np
import torch
from torch import nn
from d2l import torch as d2l
```

```python
#define methods after class has been created
def add_to_class(Class):
    """Register functions as methods in class"""
    def wrapper(obj):
        setattr(Class, obj.__name__, obj)
    return wrapper
```

```python
#goal is to implement class A with a "do" method
class A:
    def __init__(self):
        self.b = 1
a = A() 

#now use add_to_class to add to A's scope
```

```python
@add_to_class(A)
def do(self):
    print("Class atribute b is", self.b)

a.do() #just makes code more readable and not use class defintions
```

```python
#class for saving class arguments in init method as class attributes
class HyperParameters: #initial class
    """Base Params"""
    def save_hyperparameters(self, ignore=[]):
        raise NotImplemented

#create class that inherits it and calls save_hyperparams
class B(d2l.HyperParameters):
    def __init__(self, a, b, c):
        self.save_hyperparameters(ignore=['c'])
        print(self.a, self.b)
        print("There is no self.c", not hasattr(self, 'c'))
        
b = B(a=1, b=2, c=3)
```
