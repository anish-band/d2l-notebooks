```python
#run multiple minbatches gradient descent at the same time, do it by vectorizing calculations
import torch
import time

n = 1000
a = torch.ones(n)
b = torch.ones(n)
c = torch.zeros(n)

t = time.time()
for i in range(n):
    c[i] = a[i] + b[i]
print(f'{time.time() - t:.5f} sec') # slow version iterates 1000 times and adds each time (overhead per call)
```
