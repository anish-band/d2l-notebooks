```python
import torch

x = torch.tensor([2.0], requires_grad=True)  # 1. mark what you want derivative w.r.t.
y = x ** 2                                    # 2. y = x^2 at x = 2
y.backward()                                  # 3. compute gradients (goes backwards and does chain rule automatically)
print(x.grad)                                 # 4. access result → tensor([4.])
```

```python
for epoch in range(100):
    x.grad.zero_()  #0 to reset gradient so that next backwards walk has fresh start
    y = x ** 2
    y.backward()
```
